```python
class Cart:
    def __init__(self):
        self.items = {}   

    def add_item(self, name, price):
        self.items[name] = price
        print(f"Added: {name} (Rs. {price})")

    def remove_item(self, name):
        if name in self.items:
            del self.items[name]
            print(f"Removed: {name}")
        else:
            print(f"{name} not found in cart")

    def total_price(self):
        total = sum(self.items.values())
        print(f"Total: Rs. {total}")
        return total



cart = Cart()
cart.add_item("Milk",   120)
cart.add_item("Bread",  80)
cart.add_item("Butter", 200)
cart.remove_item("Bread")
cart.total_price()  
```

```python
print("Muhammad Asim f24-1700")
class ATM:
    def __init__(self, pin, balance=0):
        self.__pin     = pin      # private attribute (double underscore)
        self.__balance = balance

    def __verify_pin(self, pin):
        return pin == self.__pin

    def check_balance(self, pin):
        if self.__verify_pin(pin):
            print(f"Balance: Rs. {self.__balance}")
        else:
            print("Wrong PIN!")

    def deposit(self, pin, amount):
        if self.__verify_pin(pin):
            self.__balance += amount
            print(f"Deposited Rs. {amount}. Balance: Rs. {self.__balance}")
        else:
            print("Wrong PIN!")

    def withdraw(self, pin, amount):
        if not self.__verify_pin(pin):
            print("Wrong PIN!")
        elif amount > self.__balance:
            print("Insufficient funds!")
        else:
            self.__balance -= amount
            print(f"Withdrawn Rs. {amount}. Balance: Rs. {self.__balance}")


# --- usage ---
atm = ATM(pin=1234, balance=5000)
atm.check_balance(1234)       # Balance: Rs. 5000
atm.deposit(1234, 1000)       # Deposited Rs. 1000
atm.withdraw(9999, 500)       # Wrong PIN!
atm.withdraw(1234, 500)       # Withdrawn Rs. 500
```

```python
print("Muhammad Asim f24-1700")
class Product:
    def __init__(self, name, price):
        self.name  = name
        self.price = price

    def __str__(self):
        return f"{self.name} - Rs. {self.price}"


class Order:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)
        print(f"Added to order: {product}")

    def calculate_total(self):
        total = sum(p.price for p in self.products)
        print(f"\n--- Order Summary ---")
        for p in self.products:
            print(f"  {p}")
        print(f"Total: Rs. {total}")
        return total



p1 = Product("Laptop",   85000)
p2 = Product("Mouse",    1500)
p3 = Product("Keyboard", 3000)

order = Order()
order.add_product(p1)
order.add_product(p2)
order.add_product(p3)
order.calculate_total()  
```

```python
class SRMS:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
        self.marks = []                         # Fix 2: self.marks

    def add_marks(self, sub1, sub2, sub3):      # Fix 1 & 4: proper method, no extra param
        self.marks = (sub1, sub2, sub3)

    def calculate_percentage(self):             # Fix 1: proper method
        self.percentage = sum(self.marks) / 3   # Fix 3: sum() first
        return self.percentage

    def grade(self):                            # Fix 1: proper method
        if self.percentage >= 90:
            return "A"
        elif self.percentage >= 80:
            return "B"
        elif self.percentage >= 70:
            return "C"
        elif self.percentage >= 60:
            return "D"
        else:
            return "F"

    def __str__(self):
        return (f"Name: {self.name}, Roll No: {self.roll_no}, "
                f"Marks: {self.marks}, Percentage: {self.percentage:.2f}%, "
                f"Grade: {self.grade()}")


s1 = SRMS("Asim", "F24-1700")
s1.add_marks(85, 90, 95)
s1.calculate_percentage()
print(s1)


```
