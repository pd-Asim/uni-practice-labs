```python
# what is list
list1 = [1, 2, 3, 4, 5,1,2,'asim',True,False,"asim"]
print(list1) 
print(list1[4:10]) # indexing
list1.append("python") # add element in list
print(list1)
list1.insert(10,"java") # add element in list at specific index
print(list1)
list1.remove("python") # remove element in list
print(list1)
list1.pop(7) # remove element in list at specific index
print(list1)
fruits = ["apple", "banana", "cherry"]
list1.extend(fruits) # add element of one list to another list
print(list1)
list1.reverse() # reverse list
print(list1)
list2 = [1, 2, 3, 4, 5,5,7,8,3,5,7,89,5,2,3,4,5,89,0,2]
list2.sort() # sort list
print(list2)
list2.sort(reverse=True) # sort list in reverse order
print(list2)
list2.clear() # clear list
print(list2)
```

```python
tuple1 = (1, 2, 3, 4, 5,1,1,2,'asim',True,False,"asim")
tuple2=("apple","banana","cherry","apple")
print(tuple1)
print(tuple1[4:10]) # indexing
print(tuple1+tuple2) # concatenation
print(tuple1*2) # repetition
print(tuple1.count(1)) # count element in tuple
print(tuple1.index('asim')) # index of element in tuple
print(tuple1[0]) # indexing
print(tuple1[0:5]) # slicing
print(tuple1[0:5:2]) # slicing with step
tuple3=(7,)
print(tuple3)
print(type(tuple3)) # type of tuple
print(len(tuple1)) # length of tuple
```

```python
set1 = {1, 2, 3, 4, 5,1,2,'asim',True,False,"asim"}
print(set1) # set does not allow duplicate values
set1.add("asim") # add element in set
print(set1)
set1.remove("asim") # remove element in set
print(set1)
set1.discard(5) # remove element in set without error
print(set1)
set1.pop() # remove random element in set
print(set1)
set2 = {"apple", "banana", "cherry"}
set1.update(set2) # add element of one set to another set
print(set1)
```

```python
dict1 = {"name": "ali",
          "age": 25,
            "city": "karachi"}
print(dict1) # print dictionary
print(dict1["name"]) # indexing
dict1["age"] = 30 # update value in dictionary
print(dict1)
x=values = dict1.values() # get values of dictionary
print(x)
y=keys = dict1.keys() # get keys of dictionary
print(y)
z=items = dict1.items() # get items of dictionary
print(z)
dict1.clear() # clear dictionary
print(dict1)
```

```python
def my_function():
    print("Hello World")
my_function()

def my_info(name, age):
    print("My name is " + name + " and I am " + str(age) + " years old.")
my_info("ali", 25)
my_info("asim", 21)

def math(a, b):
    return a + b
print(math(5, 10))
math(10, 20)
math(20, 30)
 
def empty_function():
    pass

def countries(name, country="Pakistan"):
    print("i am " + name + " and I live in " + country)
countries("ali")
countries("asim", "china")
countries("adnan", "usa")
countries("ahmed")

def animals(type, name):
    print("i have a " + type + " and its name is " + name)
animals("dog", "buddy")
animals(name="suzie", type="cat")
animals("rabbit", name="bunny")
animals("lili", "hamster")

def only_positional(name, age, /):
    print("My name is " + name + " and I am " + str(age) + " years old.")
only_positional("ali", 25)
only_positional("asim", 21) # this will give error because we cannot

def only_keyword(name, age, *, country):
    print("My name is " + name + " and I am " + str(age) + " years old and I live in " + country)
only_keyword(name="ali", age=25, country="Pakistan")
only_keyword("asim", 21, country="China") # this will give error because we cannot use positional arguments after keyword arguments

def arbitraryy(*args):
    print("my name is " + args[0] + " and I am " + str(args[1]) + " years old.")
    print("i have "+str(args[2])+" brothers and "+str(args[3])+" sisters.")
arbitraryy("ali", 25, 2, 3)
```

```python
def my_function():
    x="hello world"
    print(x)
my_function()

y="how are you"
def ask():
    print(y)
ask()
print(y)

q="i am fine"
print(q)
def answer():
    global q
    q="i am good"
    print(q)
answer()
print(q)
t="tool free"
def check():
    t="not tool free"
    print(t)
check()
print(t)

def outer_function():
    x="hello"
    def inner_function():
        nonlocal x
        x="world"
        print(x)
    inner_function()
    return x
print(outer_function())

o="global variable"
def outer():
    o="outer variable"
    def inner():
        nonlocal o
        o="inner variable"
        print(o)
    inner()
    return o
print(outer())
print(o)
```
