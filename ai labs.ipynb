```python
s=12
y=20
if s>y and s%2==0:
    print("s is greater than y and s is even")
elif s>y and s%2!=0:
    print("s is greater than y and s is odd")
else:
    print("s is less than y")
```

```python
i = int(input("enter a number: "))
j = int(input("enter another number: "))

if i > j:
    print("i is greater than j")
elif i == j:
    print("i is equal to j")
else:
    print("j is greater than i")
```
